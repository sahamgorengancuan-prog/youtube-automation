# Autonomous Scientific Animation Studio — MVP Notebook

One Google Colab notebook that turns a **topic** into a complete, reusable
**storyboard with cached SVG assets** and a **static scene preview**.

```text
Topic
  → Free web research (DDGS + Wikipedia + Crossref + OpenAlex + arXiv)
  → Deterministic scientific validation
  → Retention-first script (with timing)
  → Storyboard JSON
  → Scene plan (asset catalogue)
  → SVG asset generation (only what this storyboard needs)
  → Content-addressed asset cache
  → Static scene composition + storyboard preview
```

**Scope.** This is the Phase-1 foundation. It stops at the **storyboard + scene
preview**. Video rendering (animation, voice-over, music, subtitles, MP4) is
**Phase 2** and is intentionally not included — the storyboard JSON, the SVG
cache, and the `AnimationPlaceholder` fields are the hand-off contract to it.

The deliverable of every run is: `research.json`, `script.json`,
`storyboard.json`, `scene01.svg …`, `assets/`, a shared `cache/`, and
`manifest.json` — plus a zipped copy of the run.

## Engineering decisions

- **One notebook, one importable package.** Every section writes a real module
  into `src/autostudio/`. Nothing lives only in a cell, so the exact same code
  lifts into FastAPI / Docker without a rewrite. `StudioPipeline.run()` is the
  single entry point.
- **Deterministic SVG (the key choice).** The LLM never writes SVG. It only
  *names* an `asset_type` from a fixed whitelist; Python renders the vector from
  a controlled template. That makes every asset reproducible, hashable, and
  cacheable — see the SVG-method comparison in section 10.
- **Cache-first.** Web search, every LLM response, and every SVG asset are
  content-addressed. Re-running an identical stage is a cache hit, never a
  regeneration. Assets carry metadata + a reuse counter and are shared across
  topics and runs.
- **Local-first models.** Qwen (HuggingFace) is primary; OpenAI is an optional
  fallback used only if a key is present and local generation fails. No API is
  hardcoded.
- **Low GPU usage.** The GPU is used only by Qwen. Search, validation, SVG
  generation, composition, and preview run on CPU.
- **Graceful degradation.** Each generative stage has a deterministic fallback,
  so a flaky network or a missing model never aborts the run.

Each section below is independently re-runnable in order.

## 7. Script generator

Produces a retention-first narration: **hook → curiosity → scientific
explanation → escalation → ending**, broken into timed beats. The LLM writes the
beats; Python enforces the word budget and computes speech timing / estimated
duration from `words_per_second`, so the script always fits the target length.

## 10. SVG generator

### Which SVG generation method — and why

| Approach | Deterministic | Editable vectors | Style-consistent | Cost / GPU | Verdict |
|---|---|---|---|---|---|
| **LLM writes raw SVG** | ✗ (varies per run) | ✗ (messy paths) | ✗ | tokens/GPU | rejected — not cacheable |
| **Raster model → vectorize** (e.g. diffusion + potrace) | ✗ | partial | ✗ | high GPU | rejected — raster origin, heavy |
| **HF vector models** | ✗ | partial | ✗ | GPU + immature | rejected — experimental |
| **Parametric Python templates** | ✓ | ✓ | ✓ | ~0 (CPU) | **chosen** |

The LLM decides *what* object a scene needs (`asset_type` from a whitelist); a
controlled Python template renders *how* it looks. This gives byte-identical,
reproducible, editable, flat-vector output with **no gradients, textures, or
raster** — and it is cacheable by content hash. It also matches the reference
"scientific simulation" look and is the cleanest bridge to Motion Canvas /
Remotion, which want clean editable vectors, not model output.

Only the assets the current storyboard requires are generated; everything else

Choose one of three modes:

and no model**. Also proves cache reuse and SVG safety. Useful in CI and for
verifying a fresh environment before spending GPU time.

````python
from xml.etree import ElementTree as ET

from autostudio.config import load_config
from autostudio.research import ResearchService, ScientificValidator
from autostudio.script_generator import ScriptGenerator
from autostudio.storyboard_generator import StoryboardGenerator
from autostudio.scene_planner import ScenePlanner
from autostudio.cache import AssetCache
from autostudio.composer import SceneComposer
from autostudio.schemas import ScriptBeat, ScriptPackage, SourceRecord

cfg = load_config(PROJECT_ROOT / "config" / "config.yaml")

sources = [
    SourceRecord(source_id="cro-a1", provider="Crossref", title="Rotation and atmosphere",
                 snippet="Earth rotates at about 1670 km/h at the equator.", score=0.9),
    SourceRecord(source_id="ope-b2", provider="OpenAlex", title="Ocean redistribution",
                 snippet="Stopping rotation would redistribute oceans toward the poles.", score=0.88),
    SourceRecord(source_id="arx-c3", provider="arXiv", title="Atmospheric dynamics",
                 snippet="Winds near 1700 km/h would scour the surface.", score=0.92),
]

research = ResearchService(cfg, None, PROJECT_ROOT / "cache" / "research")._fallback("Offline smoke test topic", sources)
research = research.model_copy(update={"research_hash": "smoke"})
report = ScientificValidator(cfg).validate(research)
print("validation passed:", report.passed, "| coverage:", report.coverage_score)

beats = [ScriptBeat(beat_id=f"B0{i}", purpose="beat", narration=n, evidence_refs=["F01"])
         for i, n in enumerate([
             "Imagine the Earth freezing mid spin in a single instant.",
             "The ground stops but the atmosphere keeps racing at full speed.",
             "Continent scale winds tear across every city on the planet.",
         ], start=1)]
script = ScriptGenerator(cfg, None, PROJECT_ROOT / "cache" / "scripts")._repair(
    ScriptPackage(topic=research.topic, hook="h", curiosity="c", scientific_explanation="s",
                  escalation="e", ending="end", beats=beats), research)
script = script.model_copy(update={"script_hash": "smoke"})

storyboard = ScenePlanner(cfg).plan(StoryboardGenerator(cfg, None, PROJECT_ROOT / "cache" / "storyboards")._fallback(script))
cache = AssetCache(cfg, PROJECT_ROOT / "cache")
asset_paths = {r.asset_id: cache.get_or_create(r, storyboard.topic)[0] for r in storyboard.asset_catalog}
reuse = max(cache.get_or_create(r, storyboard.topic)[1].reuse_counter for r in storyboard.asset_catalog)
assert reuse >= 1, "asset cache did not reuse"

composer = SceneComposer(cfg)
scene_dir = PROJECT_ROOT / "output" / "_smoke"
scenes = [composer.compose_scene(s, asset_paths, scene_dir / f"scene{i:02d}.svg",
                                 storyboard.canvas_width, storyboard.canvas_height)
          for i, s in enumerate(storyboard.scenes, start=1)]
for p in scenes:
    ET.parse(p)  # must be well-formed XML
    assert "lineargradient" not in p.read_text().lower()

print(f"OFFLINE SMOKE TEST PASSED — {len(scenes)} scenes, {len(asset_paths)} assets, max reuse={reuse}")
````
